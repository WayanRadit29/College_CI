merepresentasikan pengetahuan tentang Knowldge base yang berisi **orang**, **lagu**, **album**, dan **disk**.


Predikat yang tersedia:

````python
# ∀, ∃, ⇒, ∧, ∨, ¬

# Wrote(Gershwin, TheManILove)
# ¬Wrote(Gershwin, EleanorRigby)
# Wrote(Gershwin, TheManILove) ∨ Wrote(McCartney, TheManILove)
# ∀s[Sings(McCartney,s,Revolver)⇒Wrote(McCartney,s)]
# ∃d[CopyOf(d,Revolver)∧Owns(Joe,d)]
# ∀s[Sings(p,s,Revolver)⇒¬Wrote(Gershwin,s)]
# ∀s[Wrote(Gershwin,s)⇒Sings(Gershwin,s,AlbumOf(s))]
````

### **Soal 2**

Aturan dalam KB menyatakan bahwa **jika McCartney menyanyikan sebuah lagu pada album tertentu, maka Joe memiliki salinan album itu**.
Karena diketahui McCartney menyanyikan *EleanorRigby* di Revolver, buktikan dengan inference bahwa Joe memiliki salinan *Revolver*.

---

### **Soal 3**

Aturan dalam KB menyatakan bahwa **setiap lagu yang Gershwin tulis direkam di sebuah album**.
Buktikan dengan *forward chaining* dan *backward chaining* apakah lagu *EleanorRigby* (yang tidak ditulis Gershwin) juga direkam di sebuah album.



````python
#Answer here

#Soal 1
#fc
owns = list(fol_fc_ask(kb, expr('Owns(Joe, d)')))
copies = list(fol_fc_ask(kb, expr('CopyOf(d, AlbumX)')))

# cari irisan hasil
intersection = [o for o in owns if o in copies]
print(intersection)

#bc
owns = list(fol_bc_ask(kb, expr('Owns(Joe, d)')))
copies = list(fol_bc_ask(kb, expr('CopyOf(d, AlbumX)')))

intersection = [o for o in owns if o in copies]
print(intersection)



````

Hasilnya untuk forward chainning dan backward chainning itu sama-sama list kosong, kode di atas mencari intersection dari query apakah Joe punya suatu disk dan query apakah disk itu adalah salinan dari AlbumX, jika memang ada suatu disk yang merupakan salinan dari AlbumX dan dimiliki Joe maka harusna ada nilai yang dalam list seperi d : D1, karena kosong berarti tidak ada disk yang Joe punya yang merupakan salinan dari AlbumX, maka hasilnya False untuk pembuktian alias list kosong


````python
#Soal 2

#Inference menggunakan forward chainning
owns = list(fol_fc_ask(kb, expr('Owns(Joe, d)')))
copies = list(fol_fc_ask(kb, expr('CopyOf(d, Revolver)')))

# cari irisan hasil
intersection = [o for o in owns if o in copies]
print(f"Bernilai benar jika {intersection}, artinya D1 adalah copy dari revolver yang dimiliki oleh Joe sesuai KB")


````

Hasil forward chaining bernilai benar karena di dalam knowledge base terdapat dua fakta yang saling terhubung: Joe memiliki disk D1, dan D1 adalah salinan dari album Revolver. Saat inference mencari hubungan antara kepemilikan Joe dan salinan Revolver, ditemukan bahwa variabel d pada kedua fakta sama-sama bernilai D1. Artinya, disk D1 yang dimiliki Joe adalah salinan Revolver, sehingga pernyataan tersebut terbukti benar.


````python
print(list(fol_fc_ask(kb, expr('Sings(Gershwin, The_Man_I_Love, a)'))))
print(list(fol_fc_ask(kb, expr('Sings(Gershwin, Eleanor_Rigby, a)'))))


print("--------Pembatas-------")

print(list(fol_bc_ask(kb, expr('Sings(Gershwin, The_Man_I_Love, a)'))))
print(list(fol_bc_ask(kb, expr('Sings(Gershwin, Eleanor_Rigby, a)'))))
````


Hasil inferensi menunjukkan bahwa untuk lagu *The Man I Love*, baik forward chaining maupun backward chaining berhasil menemukan bahwa lagu tersebut direkam di sebuah album karena Gershwin menulis lagu itu. Sementara untuk *Eleanor Rigby*, keduanya tidak menghasilkan kesimpulan karena di dalam knowledge base terdapat fakta bahwa Gershwin tidak menulis lagu tersebut. Akibatnya, aturan “setiap lagu yang ditulis Gershwin direkam di sebuah album” tidak berlaku, sehingga *Eleanor Rigby* tidak dapat dibuktikan direkam di album mana pun.
